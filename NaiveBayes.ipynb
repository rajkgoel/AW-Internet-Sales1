# Naive Bayes

Naive Bayes is a probabilistic classification algorithm with a strong independence assumption that is used to predict the category of data point using probability. It is based on Bayes' Theorem, a mathematical formula for calculating conditional probabilities.

**Why is it "Naive"?**

The algorithm is called "naive" because it makes the strong, and often unrealistic, assumption that all features are independent of each other. For example, a fruit may be considered an apple if it is red, round, and about 3 inches in diameter; Naive Bayes assumes each of these features contributes independently to the probability of it being an apple, regardless of any actual correlations between color and shape.

**Core Assumptions**

Feature Independence: Each feature does not affect any other feature.
Equal Importance: All features contribute equally to the final prediction.


Advantages & Disadvantages
---
| Pros |	Cons |
| --- | --- |
|Fast & Efficient: Extremely quick to train and make predictions, even on large datasets. |	Unrealistic Assumption: Independence is rarely true in real-world data, which can reduce accuracy if features are highly correlated. |
|Handles High Dimensions: Works effectively for tasks with many features, like text classification. |	Zero Frequency Problem: If a category appears in test data but not in training, the model assigns it a zero probability (often fixed with Laplace smoothing). |
|Works with Small Data: Can perform well even with limited training examples.	| Poor Estimator: While good at classifying, its probability outputs are often not highly reliable. |


Real-World Applications
---
- Spam Filtering: Identifying unwanted emails based on keywords like "free" or "winner".
- Sentiment Analysis: Categorizing reviews or social media posts as positive, negative, or neutral.
- Medical Diagnosis: Assisting in predicting diseases based on symptoms and patient history.
- Recommendation Systems: Helping suggest products or content by predicting user preferences.

Core Idea
---

Naive Bayes calculates:
>  P(Class ∣ Features)

Using
> $ P(A|B) = P(B|A)P(A) / P(B) $

This is Bayes' Theorem.

**Formula Used in Classification**

We compute for each class:
> $ P(Class)×P(Feature_1 ​∣ Class) × P(Feature_2 ​∣ Class)×.. $

Then choose the class with highest probability.



Types of Naive Bayes in scikit-learn

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# 1️⃣ Create Dataset
data = {
    "Age": [25, 40, 35, 28, 50, 45, 23, 38, 29, 48],
    "Income": [30000, 80000, 60000, 35000, 90000, 75000, 25000, 65000, 40000, 85000],
    "Marital_Status": ["Single", "Married", "Married", "Single",
                       "Married", "Married", "Single", "Single",
                       "Single", "Married"],
    "Loan_Approved": [0, 1, 1, 0, 1, 1, 0, 1, 0, 1]
}

df = pd.DataFrame(data)

# Encode categorical feature
le = LabelEncoder()
df["Marital_Status"] = le.fit_transform(df["Marital_Status"])

# Split features and target
X = df[["Age", "Income", "Marital_Status"]]
y = df["Loan_Approved"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Train Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)

# Predict
y_pred = nb.predict(X_test)
print("Test Data: \n", X_test)
print("-" * 30)
print("Prediction: ", y_pred)
print("-" * 30)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Spam email detection
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# 1️⃣ Create Sample Dataset
data = {
    "email": [
        "Win a free iPhone now",
        "Limited time offer claim prize",
        "Meeting scheduled tomorrow",
        "Project discussion at 5pm",
        "Congratulations you won lottery",
        "Let's have lunch today",
        "Earn money quickly from home",
        "Please review the attached document"
    ],
    "label": [1, 1, 0, 0, 1, 0, 1, 0]  # 1 = Spam, 0 = Not Spam
}

df = pd.DataFrame(data)

# 2️⃣ Split into train & test
X_train, X_test, y_train, y_test = train_test_split(
    df["email"], df["label"], test_size=0.3, random_state=42
)

# 3️⃣ Convert text into numerical features
vectorizer = TfidfVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# 4️⃣ Train Naive Bayes model
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# 5️⃣ Predict
y_pred = model.predict(X_test_vec)

print("Test Data: \n", X_test)
print("-" * 30)
print("Prediction: ", y_pred)
print("-" * 30)

# 6️⃣ Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

new_email = ["Congratulations you won cash prize"]
new_email_vec = vectorizer.transform(new_email)
prediction = model.predict(new_email_vec)

print("Spam" if prediction[0] == 1 else "Not Spam")
```
